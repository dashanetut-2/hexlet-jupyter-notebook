```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r'C:\Users\adminadmin\Downloads\Global_Education_2.csv', encoding = "latin1")

df.info()

print(df.head())

df_plot = df.nlargest(
    10, 'Gross_Tertiary_Education_Enrollment'
).sort_values('Gross_Tertiary_Education_Enrollment')

df_plot.plot(
    x='Countries and areas',
    y='Gross_Tertiary_Education_Enrollment',
    kind='bar',
    figsize=(10, 6),
    legend=False
)

plt.title('Топ-10 стран по охвату высшим образованием')
plt.xlabel('Страна')
plt.ylabel('Охват высшим образованием (%)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
